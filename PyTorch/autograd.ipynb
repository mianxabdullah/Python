```python
import torch

#require_grad = True/False is a flag that tells PyTorch to track all operations on the tensor. 
# When you set require_grad = True, it allows pytorch to change the tensor in future to improve the model.


#Computational graph will only be created when require_grad = True. 
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
y= x * 2
y
```

```python
# Backpropagation is the process of computing the gradients of the loss function with respect to the model's parameters.

#example of backpropagation
a=torch.tensor([2.])
b=torch.tensor([5.], requires_grad=True)

y_pred=a*b
y_true=torch.tensor([12.])

loss=(y_pred-y_true)**2
print("Loss:", loss.item())
print("Actual value:", y_true.item())
print("Predicted value:", y_pred.item())
loss.backward() # compute the gradients of the loss with respect to the model's parameters

print("Gradient of b:", b.grad.item()) # print the gradient of b

# as gradient of b is 2*(a*b-y_true)*a = 2*(10-12)*2 = -8
# now we know its negative so we need to increase the value of b to reduce the loss.
```

```python
"""
.grad stores the instructions on how to change a number to reduce the loss. 
It is used in optimization algorithms like gradient descent to update the model's parameters and improve its performance.

.grad tells 2 things:
1. The direction to change the number to reduce the loss.
2. The magnitude of change needed to reduce the loss.

if w.grad > 0, we need to increase the value of w to reduce the loss.
if w.grad < 0, we need to decrease the value of w to reduce the loss.
"""
```

```python
# Gradient descent is an optimization algorithm used to minimize the loss function by iteratively updating the model's 
# parameters in the direction of the negative gradient of the loss function with respect to the parameters.

#Formula: new_weight = old_weight - learning_rate * gradient

# input fixed
x=torch.tensor((2.0))

# learnable parameter
w=torch.tensor((5.0), requires_grad=True)

#correct output
y_true=torch.tensor((12.0))

#learning rate
lr=0.1

y_pred=x*w
loss=(y_pred-y_true)**2

loss.backward() # compute the gradients of the loss with respect to the model's parameters

print("Before update") # print the value before update
print("w:", w.item()) # print the value of w
print("Loss:", loss.item()) # print the loss
print("Gradient of w:", w.grad.item()) # print the gradient of w

with torch.no_grad(): # we don't want to track the gradients while updating the weights
    w -= lr * w.grad # update the weights using gradient descent

w.grad.zero_() # reset the gradients to zero after updating the weights

print("After update") # print the value after update
print("w:", w.item()) # print the value of w
print("Loss:", loss.item()) # print the loss
print("Gradient of w:", w.grad.item()) # print the gradient of w
```

```python
# Training loop

x=torch.tensor([2.0,4.0,6.0,8.0])
y_true=torch.tensor([4.0,8.0,12.0,16.0])

w=torch.tensor([1.0], requires_grad=True)
lr=0.01
epochs=10

for epoch in range(epochs):
    # step 1: make predictions
    y_pred=x*w

    #step 2: calculate the loss
    loss=((y_pred-y_true)**2).mean()

    #step 3: backpropagate the loss
    loss.backward() 

    #step 4: update the weights using gradient descent
    with torch.no_grad():
        w -= lr * w.grad # update the weights using gradient descent
        w.grad.zero_()

    print(f"Epoch {epoch+1}: Loss: {loss.item():.4f}, w: {w.item():.4f}") # print the loss and the value of w
```
