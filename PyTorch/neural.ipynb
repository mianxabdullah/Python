```python
# Class Structure
import torch
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()

        # A learnable parameter
        self.w = nn.Parameter(torch.tensor(1.0))

    def forward(self, x):
        return x * self.w


# Create the model
model = MyModel()

# Give it an input
x = torch.tensor(5.0)

# Get the prediction
output = model(x)

print(output)
print(model.w)
```

```python
# Activation Functions: they introduce non-linearity into the model, allowing it to learn complex patterns. AF always comes after layers
# Common activation functions include ReLU, Sigmoid, and Tanh.

# ReLU (Rectified Linear Unit) is a popular activation function that outputs the input directly if it is positive; otherwise, it will output zero.
# sigmoid squashes the input to a range between 0 and 1, making it useful for binary classification problems.
# softmax is often used in the output layer of a classification model to convert raw scores (logits) into probabilities that sum to 1.

#ReLU is used in hidden layers, while softmax is used in the output layer for multi-class classification problems. 
# and sigmoid is used in the output layer for binary classification problems.
```

```python
# Optimizers: they are algorithms that adjust the weights of the model based on the gradients computed during backpropagation.
# Some common optimizers include Stochastic Gradient Descent (SGD), Adam, and RMSprop.
# SGD updates the weights by moving them in the direction of the negative gradient of the loss function with respect to the weights.
# Adam is an adaptive learning rate optimization algorithm that combines the advantages of both AdaGrad and RMSprop.
# RMSprop is an optimizer that adapts the learning rate based on the average of recent squared gradients.

import torch
import torch.nn as nn
import torch.optim as optim

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.Linear = nn.Linear(1, 1) # A linear layer with 1 input feature and 1 output feature

    def forward(self, x):
        return self.Linear(x)

x=torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y=torch.tensor([[2.0], [4.0], [6.0], [8.0]])

model = MyModel()

loss_fn = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

epochs = 10
for epoch in range(epochs):
    optimizer.zero_grad()  # Zero the gradients
    y_pred = model(x)  # Forward pass
    loss = loss_fn(y_pred, y)  # Compute the loss
    loss.backward()  # Backward pass
    optimizer.step()  # Update the weights

    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")

# Save the model
torch.save(model.state_dict(), "model.pth")
```
