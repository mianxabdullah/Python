```python
import seaborn as sns
import pandas as pd

data = pd.read_csv('ecommerce_sales.csv')
data["order_date"] = pd.to_datetime(data["order_date"])
data["year"] = data["order_date"].dt.year
data.head()
```

```python
sns.countplot(x='region',data=data,order=data['region'].value_counts().index)
```

```python
# there is no pie chart in seaborn, but we can use matplotlib to create a pie chart.
```

```python
# histogram:
sns.histplot(data['customer_rating'],bins=10)
```

```python
# KDE (Kernel Density Estimation) plot: smooth curve but same as histogram, but it is a continuous probability distribution function that estimates the probability density function of a random variable.
# used where you want to plot distribution of data points across different value ranges, but you want to see a smooth curve instead of bars. 
# more density= smoother curve, less density= more jagged curve.
sns.kdeplot(data['customer_rating'])
#skewness: if the histogram is skewed to the left, it means that the data is negatively skewed, if the histogram is skewed to the right, it means that the data is positively skewed, if the histogram is symmetric, it means that the data is normally distributed.
# example: left long tail = negatively skewed, right long tail = positively skewed, bell shaped = normally distributed
# in left and right skewed distributios, mean isnt equal to median and it isnt equal mode, but in normal distribution, mean=median=mode.
```

```python
sns.lineplot(x='year',y='revenue',data=data)
```
